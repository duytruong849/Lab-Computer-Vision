```python
#2. Đọc và hiển thị ảnh
from PIL import Image, ImageDraw, ImageFont
img = Image.open('images.jpg')
img.save('images.png')
```

```python
#3. Chuyển đổi không gian màu
img = Image.open('images.jpg')

img_grayscale = img.convert("L")
img_grayscale.save('images_graysacled.jpg')

img_HSV = img.convert("HSV")
h,s,v = img_HSV.split()
h.save('img_HSV_h.png')
s.save('img_HSV_s.png')
v.save('img_HSV_v.png')

img_LAB = img.convert("LAB")
img_LAB.save('images_LAB.tif')
```

```python
#4. Cắt xén và thay đổi kích thước
img = Image.open('images.jpg')
width, height = img.size
print(width, height)
img_sliced = img.crop((60,100,6000,300))
img_sliced.save('images_sliced.jpg')

img_resized = img.resize((800,600))
img_resized.save('images_resized.jpg')
```

```python
#5. Vẽ hình cơ bản
import numpy as np
img = Image.open('images.jpg')

img_draw = ImageDraw.Draw(img)

img_draw.circle((100,200),50,'yellow')
img_width, _ = img.size
img_draw.line((0,200,img_width,200),'red')
img_draw.rectangle((200, 150, 250, 250),'blue')
img.save('img_drawed.jpg')

img = Image.open('images.jpg')
img_text = ImageDraw.Draw(img)

font = ImageFont.truetype("arial.ttf", 100)
img_text.text((150, 200),"Hello World!",fill="red",font=font)
img.save('img_texted.jpg')
```
